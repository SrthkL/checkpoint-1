```python
import pandas as pd
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt

# Sample DataFrame for demonstration

# Calculate mean, median, and mode for each numeric column
mean_values = df1.mean()
median_values = df1.median()
mode_values = df1.mode().iloc[0] # Take the first mode in case of multiple modes

# Combine the statistics into a single DataFrame for visualization
stats_df =pd.DataFrame({
    'Mean': mean_values,
    'Median': median_values,
    'Mode': mode_values
})

# Plot the statistics
stats_df.plot(kind='bar', figsize=(8, 6))
plt.title('Mean, Median, and Mode for Each Column')
plt.ylabel('Values')
plt.xticks(rotation=0)
plt.legend(title='Statistics')
plt.tight_layout()
plt.show()
```
